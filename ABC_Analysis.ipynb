```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python


df = pd.read_excel('/content/drive/MyDrive/ABC_Company.xlsx')
print(df.head())
df.head()
```

```python
print(df.columns)
```

```python
np.random.seed(42)
df['height'] = np.random.randint(150, 181, size=len(df))
df.head()
```

```python
team_count = df['Team'].value_counts()
team_percentage = (team_count / len(df)) * 100

print(team_count)
print(team_percentage)
```

```python
plt.figure()
team_count.plot(kind='bar')
plt.title("Employee Distribution by Team")
plt.xlabel("Team")
plt.ylabel("Number of Employees")
plt.show()

```

```python
position_count = df['Position'].value_counts()
print(position_count)
```

```python
plt.figure()
position_count.plot(kind='bar')
plt.title("Employees by Position")
plt.xlabel("Position")
plt.ylabel("Count")
plt.show()
```

```python
bins = [20,25,30,35,40,45,50]
labels = ['20-25','26-30','31-35','36-40','41-45','46-50']

df['AgeGroup'] = pd.cut(df['Age'], bins=bins, labels=labels)

age_group_count = df['AgeGroup'].value_counts()
print(age_group_count)
```

```python
plt.figure()
age_group_count.plot(kind='pie', autopct='%1.1f%%')
plt.title("Predominant Age Group")
plt.ylabel("")
plt.show()
```

```python
team_salary = df.groupby('Team')['Salary'].sum()
position_salary = df.groupby('Position')['Salary'].sum()

print(team_salary)

print(position_salary)
```

```python
plt.figure()
team_salary.plot(kind='bar')
plt.title("Salary Expenditure by Team")
plt.ylabel("Total Salary")
plt.show()

plt.figure()
position_salary.plot(kind='bar')
plt.title("Salary Expenditure by Position")
plt.ylabel("Total Salary")
plt.show()
```
